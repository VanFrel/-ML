## § 1. О базовых идеях разведочного анализа

Exploratory Data Analysis (EDA), или разведочный анализ данных, — это подход к анализу данных, нацеленный на первичное изучение набора данных: выявление его основных характеристик, закономерностей, аномалий и взаимосвязей между переменными. EDA выполняется до построения сложных моделей и проверки гипотез.

Концепцию EDA разработал американский математик Джон Тьюки (John Tukey) в 1970‑х годах. Его идея: «Изучай данные, прежде чем их моделировать».

### Почему нельзя сразу перейти к моделям

Вот метафора, простенькая, но вполне себе точная: приходим к врачу, а он, не спрашивая о симптомах, не глядя на анализы, сразу назначает сложнейшую операцию, статистика выживаемости которой — 50 на 50. Не правда ли, хочется убежать и больше не возвращаться в медучреждение, где работают такие «специалисты»? Вот так же абсурдно в Data Science — сразу строить модели, не посмотрев на данные.

Разведочный анализ — это знакомство с данными. Мы смотрим: «Что это? Что там внутри? Есть ли пропуски? Выбросы? Как распределены признаки? Нет ли между ними скрытых связей?» Модели — это сложно. Они требуют дисциплины. Но если данные грязные, модель натренируется на мусоре — и выдаст мусор. Trash in, trash out. EDA — это предварительный шаг, который всегда предшествует построению моделей.

### Из чего состоит разведочный анализ

EDA — это набор очень простых, но очень мощных инструментов. Не обязательно сразу применять сложные алгоритмы, иногда достаточно просто нарисовать картинку. Основные приёмы разведочного анализа таковы.


#### Статистические сводки

- Среднее, медиана, стандартное отклонение, квартили, минимум, максимум — для числовых признаков.
- Таблицы частот — для категориальных признаков.
- Проверка на пропуски.


#### Очистка и первичное преображение

- Заполнение или удаление пропусков.
- Выявление и обработка или удаление выбросов.


#### Визуализация

- Гистограммы — чтобы понять распределение каждого признака.
- Box plots (ящики с усами) — чтобы увидеть выбросы и медиану.
- Scatter plots (диаграммы рассеяния) — чтобы увидеть взаимосвязь между двумя признаками.
- Тепловые карты корреляций — спорно, но иногда бывает полезно.

#### Генерация гипотез

- Нпример: «А не связан ли возраст клиента с вероятностью оттока?»
- Или: «Не различаются ли продажи по регионам статистически значимо?»

**Вывод:** EDA — это не про «посмотреть», это про «подумать» и сформулировать вопросы.

### Антитеза разведочного анализа

Первая мысль, которая приходит в головыу: «Если есть разведочный анализ, то, наверное, есть какой-то другой анализ, не являющийся разведочным?» Хороший вопрос. Ответ — не в противопоставлении «EDA vs Not EDA», а в более широкой дихотомии.

Если EDA — это гибкий, исследовательский, визуальный, непараметрический, человеко-ориентированный подход к данным, то его антитеза — это Confirmatory Data Analysis (CDA) — подтверждающий анализ. CDA — это когда у нас уже есть гипотеза, есть план, и мы проверяем её с помощью статистических тестов (t-test, ANOVA, p-values и т. д.). Это формальный, жёсткий, ориентированный на принятие решений подход. Разница примерно такова.

- **EDA**: «Посмотрим, что тут у нас интересного».
-  **CDA**: «Гипотеза H0: средние равны. Проверяем при alpha = 0.05».

Они не анотогинисты. скорее, они — две фазы одного процесса. Сначала EDA — чтобы найти. Потом CDA — чтобы доказать, что ты что-то нашёл.

Все машинное обучение — это своеобразная пирамида. В основании — EDA: мы собираем информацию, рисуем, задаём дурацкие вопросы, смотрим на выбросы и т. д. В середине — Feature Engineering: мы создаём новые признаки, чистим данные, готовим их к моделям. На вершине — моделирование и валидация. И всё это пронизано любопытством и здоровым скептицизмом.

### Две парадигмы анализа данных

Есть вещи, о которых методологи Data Science спорят уже лет десять. Дело в том, что есть две парадигмы анализа данных.


Первая — классическая, **машинно-ориентированная**. Она состоит в том, что EDA — это прелюдия. Это «разведка». Само название намекает, что главное — впереди: там будет модель, прогноз, автоматизация, и вообще будет круто. В этой парадигме EDA — служанка.


Вторая — аналитическая, **человеко-ориентированная**. В ней EDA дает самодостаточный результат. Мы собираем данные, визуализируем, вычисляем, интерпретируем — и на выходе получаем знание, а не алгоритм. И это знание можно доложить совету директоров. На его основе можно принять решение (открыть новый магазин здесь, а не там; сменить поставщика; запустить рекламную кампанию на эту аудиторию, а не на ту).

Исли мы придержживаемся второй парадигмы, то название «разведочный анализ» звучит как-то несолидно. «Разведочный» — как будто мы ещё не решили, как будто мы только ищем, а решение будет потом. Но на практике в Business Intelligence, в аналитике продукта, в маркетинговых исследованиях — EDA и есть финал.


Что с этим делают умные люди? — правильно: они перестают называть это разведочным анализом данных в серьёзных отчётах. Они говорят:

- «Аналитическое исследование данных»,
- «Дескриптивный анализ»,
- «Исследовательский анализ»,
- или просто: «Анализ данных» — и это звучит достойно.

Но внутри — это тот же самый EDA. Та же гистограмма. Тот же box plot. Та же таблица частот. Только отношение другое: не «сейчас посмотрим, а потом сделаем», а «мы уже сделали — мы поняли все, что нам нужно».

### EDA + CDA = самостоятельный инструмент

Именно так это и работает в реальном бизнесе. EDA ставит вопросы, CDA проверяет, не обманывают ли нас случайность и шум, а вместе они образуют замкнутый контур:

1) посмотрели (провели EDA — получили гипотезу),
2) проверили (провели CDA — подтвердили или опровергли гипотезу),
3) убедились (сделали вывод),
4) приняли решение (совершили управленческое действие).

И да: это самодостаточный процесс. Модели машинного обучения могут вообще не понадобиться. Более того: **в 80% бизнес-задач они не нужны**. Аналитика данных (EDA + CDA) даёт ответы быстрее, дешевле и прозрачнее.

Но что самое смешное — в академической среде часто считается, что если ты не построил модель, то ты как будто не доработал. Академики и профессора считают EDA черновой работой. А на практике именно черновая работа приносит деньги. Не нейросеть. Не градиентный бустинг. А аккуратно построенная гистограмма, на которой видно, что клиенты уходят на 30-й день. EDA — это не предварительное исследование, очень часто оно — окончательное. 

Просто человеческий язык отстаёт от реальности. Название «разведочный анализ» исторически закрепилось, и теперь мы с ним живём.

## § 3. Пропуски данных: диагностика и стратегии обработки

Это, пожалуй, один из самых важных моментов в жизни любого аналитика данных. Пропуски. Они бывают разными. Как люди. Как коты. Как настроение вашей девушки после после третьей чашки кофе (или настроение приятеля после третьей рюмки). 

Пропуски бывают:


1) **MCAR (Missing Completely At Random)** — пропуски случайные. Как если бы вы спросили у ста человек их любимый цвет, а пять человек просто не услышали вопроса. Никакой системы. Чистая случайность. С ними проще всего.

2) **MAR (Missing At Random)** — пропуски систематические, но объяснимые. Например, женщины чаще отказываются указывать вес. Или пожилые люди реже заполняют поля о физической активности. Пропуск зависит от других известных нам переменных.

3) **MNAR (Missing Not At Random)** — пропуски неслучайные и коварные. Когда сам факт пропуска несет информацию. Например, люди с низкой удовлетворенностью продуктом не ставят оценку. Или пациенты с тяжелыми симптомами пропускают визиты к врачу.

Почему это важно? Потому что неправильная обработка пропусков может исказить результаты, убить корреляции и привести к выводам, от которых настоящий аналитик будет плакать горькими слезами. Наша задача сегодня — научиться диагностировать тип пропусков и выбирать правильную стратегию. Не просто «удалить всё», а подумать: «А что стоит за этим NaN?»

Поехали.

### Датасет №1: «Косметический тест-драйв»

```python
df_cosmetic = pd.read_csv('data/df_cosmetic.csv')
df_cosmetic.info()
```

#### Поле `after_hydration` — 12% пропусков

**Тип пропусков.** MCAR (случайные). Половина участниц просто не явилась на финальный замер. Кто-то проспала. Кто-то уехала в командировку. Кто-то решила, что крем не настолько хорош, чтобы ради него вставать в 7 утра. Но в целом — случайность.

**Стратегия.** Замена на среднее по группе `'skin_type'`. Если тип кожи тоже пропущен — берем медиану по всей выборке.

**Почему.** Мы сохраняем информацию о типе кожи (жирная кожа может иметь другие показатели увлажненности, чем сухая). Это умнее, чем тупо заменить одним числом.

Сначала посмотрим, какие `'skin_type'` вообще есть в косметологии, чтобы понимать группы:

```python
df_cosmetic['skin_type'].value_counts()
```

Ок. Четыре группы: сухая, нормальная, жирная и комбинированный тип. Распределены примерно поровну.

Теперь считаем средние значения `'after_hydration'` по каждой группе `'skin_type'`:

```python
group_means = df_cosmetic.groupby('skin_type')['after_hydration'].mean()
group_means
```

Различия не то чтобы сильные — но они есть. Например, сухая кожа явно отличается от жирной.

Дальше нам понадобится функция для замены пропусков на конкретные значения. Пишем фунцию `fill_hydration`:

```python
def fill_hydration(row):
    if pd.isna(row['after_hydration']):
        if pd.isna(row['skin_type']):
            # Если и тип кожи пропущен — берем общую медиану
            return df_cosmetic['after_hydration'].median()
        else:
            # Если тип кожи известен — берем среднее по группе
            return group_means[row['skin_type']]
    else:
        # Если не пропуск — оставляем как есть
        return row['after_hydration']
```

И применяем наш любимый мтод `apply`:

```python
df_cosmetic['after_hydration'] = df_cosmetic.apply(fill_hydration, axis=1)
df_cosmetic.info()
```

**Комментарий.** Мы не просто заменили на среднее — мы сохранили логику данных. Потому что крем для сухой кожи и крем для жирной — это разные истории. Если бы мы заменили всё общей средней, мы бы смешали разные типы кожи и потеряли информацию. А так — мы молодцы.

#### Поле `satisfaction_score` — 8% пропусков

**Тип пропусков.** MNAR (неслучайные). Пропуски коррелируют с низкой удовлетворенностью. Люди, которым крем не понравился, реже ставят оценку. Если мы просто удалим эти строки — мы сделаем вид, что недовольных не существует. Если заменим средним — мы придумаем им оценку, которой не было. И то, и другое — плохо.

**Стратегия.** Создать бинарный признак `'satisfaction_missing'` (1 — пропуск, 0 — значение есть). А сам столбец `'satisfaction_score'` можно заполнить чем угодно (медианой, нулем) — мы всё равно будем использовать новый признак как индикатор.

**Почему.** Сам факт пропуска становится полезной информацией. Мы можем проанализировать: «А что, если пропуск — это на самом деле низкая оценка?»

Сначала создаем бинарный столбец-индикатор пропусков:

```python
df_cosmetic['satisfaction_missing'] = df_cosmetic['satisfaction_score'].isna().astype(int)
df_cosmetic.info()
```

```python
df_cosmetic['satisfaction_score'] = df_cosmetic['satisfaction_score'].fillna(df_cosmetic['satisfaction_score'].median())
df_cosmetic.info()
```

**Комментарий.** Теперь мы не теряем информацию. Мы говорим: «Смотрите, у нас есть два типа людей: те, кто поставил оценку, и те, кто не поставил. А не поставили они, вероятно, потому что им не понравилось. Давайте это проверим». Это называется —  извлечь сигнал из шума.

После всего этого можно удалить вспомогательный столбец `'satisfaction_missing'`, который мы специально создали для заполнения пропусков (хотя, можно и не удалять, так как он еще может понадобиться в будущем). Удаляем:

```python
df_cosmetic = df_cosmetic.drop(columns = ['satisfaction_missing'])
df_cosmetic
```

#### Поле `'skin_type'` — 5% пропусков

**Тип пропусков.** MCAR (случайные). Просто забыли заполнить поле в анкете. Никакой системы. Бывает.

**Стратегия.** Удаление строк.

## § 4. Охота на выбросы: как отличить ошибку от закономерности

В предыдущем параграфе мы навели порядок с пропусками — заполнили, удалили, где надо. Но чистый датасет — это ещё не рабочий датасет. Потому что в численных признаках прячутся выбросы: значения, которые выбиваются из общего ряда.

С ними есть две проблемы:

- они могут быть ошибками ввода — и тогда их нужно удалять,
- они могут быть редкими, но реальными случаями — и тогда их нужно сохранять, но учитывать при анализе.

Как отличить одно от другого? Для этого есть метод IQR (межквартильный размах) и визуальный анализ. Сейчас мы пройдёмся по всем трём датасетам, найдём выбросы и решим, что с ними делать.

### Общий инструментарий

Основным инструментом выявления выбросов на этом этапе анализа данных является метод интерквартильного размаха. 
Интерквартильный размах (Interquartile Range, IQR) — это мера статистического разброса данных, равная разнице между третьим Q3 и первым Q1 квартилями распределения: IQR=Q3−Q1

Квартили делят упорядоченный набор данных на четыре равные части:

- Первый квартиль Q1 — значение, ниже которого находится 25 % наблюдений (25‑й перцентиль).

- Второй квартиль Q2 — медиана, делит данные пополам (50‑й перцентиль).

- Третий квартиль Q3 — значение, ниже которого находится 75 % наблюдений (75‑й перцентиль).

Таким образом, IQR охватывает центральные 50 % данных — от 25‑го до 75‑го перцентиля.

Ключевые преимущества IQR:

- Устойчивость к выбросам. В отличие от размаха (max−min) или стандартного отклонения, IQR не зависит от экстремальных значений на краях распределения.

- Подходит для несимметричных распределений. Не требует нормальности данных.

- Наглядность. Легко интерпретируется как диапазон «типичных» значений (центральные 50 %).

Для выявление выбросов в данных применяется стандартная методика: в качестве нижнего порога полагают −1,5×IQR, в качестве верхнего прога полагают +1,5×IQR, а все наблюдения за этими границами считаются выбросами. Множитель 1,5 — эмпирический стандарт, но может меняться (например, на 2,0 или 3,0 для более строгих/мягких критериев).

Заметим, что при построении `boxplot` (ящика с усами) границы «ящика» соответствуют Q1 и Q3, а «усы» — рассчитанным порогам выбросов.

### Датасет №1: «Косметический тест-драйв»

```python
df_cosmetic.info()
```

#### Выбор признаков для анализа

Выбираем все числовые столбцы (в категориальных столбцах не может быть выбросов), которые могут нам понадобиться:

```python
cols = ['age', 'baseline_hydration', 'after_hydration', 'satisfaction_score', 'days_used']
df = df_cosmetic.copy()
```

#### IQR

И прогоняем их в цикле через интерквартильный размах:

```python
for col in cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    print(col, ':', len(outliers), 'выбросов')
    print(outliers[col])
    print('')
```

#### Вердикт

- Выбросы по возрасту — удаляем (столько люди не живут).
- Выбросы по базовой гидрации — удаляем нулевые значения (кожа не может быть абсолютно сухой).
- Выбросы по повторной гидрации — удаляем 185 (явно выбивается из остального ряда).

Выполняем локализацию по маске и сбрасываем индекс:

```python
mask =  ((df_cosmetic['age'] == 243) | (df_cosmetic['age'] == 156) | 
         (df_cosmetic['baseline_hydration'] == 0 )| 
         (df_cosmetic['after_hydration'] == 185.0))

df_cosmetic = df_cosmetic.loc[~mask]
df_cosmetic = df_cosmetic.reset_index(drop=True)
df_cosmetic.info()
```

### Краткое резюме

- IQR — это разведка. Он подсвечивает позиции. Он не выносит приговор.

- Ни одно значение не удаляется без ответа на вопрос: «Почему это значение здесь оказалось?» Ошибка ввода? Патология? Артефакт измерения? Экстремальный, но реальный случай?

- На каждый признак нужен свой подход. Нельзя сказать «удаляем все значения за 1.5 IQR». Это аналитическое преступление.

Аналитик данных не работает в вакууме. Он работает с людьми — с врачами, с биологами, с инженерами. И его задача — не «удалить выбросы», а понять, что стоит за этими числами. Потому что за каждым числом — какой-то конктретный случай. Например, пациент. Реальный, живой человек. А он не выброс. Он просто другой.

## § 5. Логика датасайентиста: как рождаются гипотезы

Когда в поиске выбросов мы гоняли признаки через IQR, мы поступали логично: брали все числовые признаки, которые являются осмысленными (id — не брали), и в цикле прогоняли их  все. Такая логика ясна: так мы проверяли абсолютно все.

Но что касается генерации гипотез: откуда берутся предположения, которые мы формулируем? Из его опыта аналитика? Из интуиции? Ведь перебирать все возможные варианты предположений — невозможно. Но ведь должна быть в его работе какая-то логика. Иначе получается, что это какое-то случайное угадывание.

Давайте разбираться. Для начала представим, что данные — это темная комната. А мы стоим на пороге с фонариком. И вот у нас возникает естественный вопрос: куда светить?

### Первый слой: структурная логика

Это самый надежный источник. Аналитик смотрит на природу данных и задает вопросы, которые вытекают из того, **что эти данные измеряют**.

- У нас есть состояния до и после применения крема. → Логично спросить: «Есть ли сдвиг?» Это не интуиция, это структура эксперимента.

- У нас есть расстояние и время доставки. → Логично спросить: «Зависит ли время от расстояния?» Потому что иначе зачем мы мерим расстояние?

- У нас есть давление и BMI. → Логично спросить: «Растет ли давление с весом?» Это медицинский факт, не догадка.

Первый раунд гипотез рождается **из смысла данных**. Мы смотрим на названия признаков и говорим: «Если я измеряю X и Y, я обязан проверить их связь, иначе я не делаю свою работу».

Это — обязательный минимум. Его можно (и нужно) формализовать в чек-лист.

### Второй слой: статистическая логика

После того как доменные гипотезы проверены, аналитик смотрит на **отдельные распределения** и задает вопросы, вытекающие из того, как они выглядят.

- В гистограмме возраста я вижу два горба. → Гипотеза: «Может, у нас две разные популяции? Молодые и пожилые? Или мужчины и женщины имеют разное распределение по возрасту?»

- В box plot удовлетворенности я вижу выбросы. → Гипотеза: «Кто эти люди? Может, они и крем не использовали? Или, наоборот, у них аллергия?»

- Я вижу, что холестерин у женщин имеет больший разброс, чем у мужчин. → Гипотеза: «Может, среди женщин есть подгруппа с особым риском?»

Этот слой — переход от «что я знаю» к «что данные мне говорят». Мы не гадаем, а **читаем графики**. Каждый артефакт на графике — это потенциальная гипотеза.

### Третий слой: комбинаторная логика

Здесь мы начинаем перебирать **комбинации признаков**, руководствуясь логикой связности.

Что у нас в принципе может быть:

- числовой признак + числовой → например, scatter.

- категориальный + числовой → boxplot, хотя, возможны и другие варианты.

- категориальный + категориальный → это не так интересно, но можно посмотреть на heatmap.

Важно: мы перебираем **не все пары, а только те, которые имеют смысл**. Но — и это тоже важно — мы перебираем **все комбинации смысловых пар**.


Алгоритм для аналитика:

1) Выписать все числовые признаки.

2) Выписать все категориальные.

3) Построить матрицу: числовой × числовой для всех пар, которые имеют смысл в предметной области.

4) Построить матрицу: категориальный × числовой для всех пар, где категориальный может объяснять разброс числового.

5) Построить матрицу: категориальный × категориальный для всех пар, где можно ожидать связь.


Это — системный перебор, и **он отнюдь не бесконечный**. Например, если у нас 5 числовых и 3 категориальных признака, мы должны будем рассмотреть 10 scatter-ов, 15 boxplot-ов и 3 (или 6, если нет симметричности) кросс-таблиц heatmap. Это вменяемое число графиков.

### Главное — это не угадывание

Датасайентист не угадывает. Датасайентист рассуждает структурно.


Он не говорит: «А давай проверим, может, связаны количество букв в комментарии и цвет волос?» — это угадывание.
Он говорит: «У меня есть control и experiment (до/после) — я обязан проверить эффект. У меня есть категориальный признак (регион) и числовой (время доставки) — я обязан проверить, различаются ли группы. У меня есть два числовых (расстояние и время) — я обязан проверить наличие связь».

EDA — это рабочая дисциплина. Она превращает Data Science **из эфемерного искусства в технологию**: мы не ждем вдохновения, мы просто знаем, что делать. Итак, друзья мои, откуда берутся гипотезы в EDA?

1) Из смысла данных (структурная логика) — обязательно.

2) Из формы распределений (статистическая логика) — читаем графики.

3) Из комбинаций признаков (комбинаторная логика) — системно, но не бесконечно.

4) Дополнительно: из опыта и контекста (экспертная логика) — если знаем предмет.

Визуализация — это не «А давайте рисовать все подряд, и посмотрим, что получится». Визуализация — это инструмент проверки гипотез. Сначала гипотеза, потом график.

## §6. Визуализация гипотез: от структурной логики к CDA

До этого момента мы работали с визуализацией как с инструментом описательным. Строили распределения, смотрели на разбросы, вылавливали выбросы и пропуски. Это было честно, полезно и скучновато, если уж говорить начистоту.

Теперь игра меняется. Мы превращаем график в инструмент проверки гипотез.Каждый график становится частью рассуждения: «Я думаю, что в данных есть связь — вот график, который её показывает (или опровергает)».  

Порядок действий в этом параграфе таков.

1) Формулируем гипотезу. Желательно — вслух и с выражением.

2) Строим визуализацию — ровно одной строкой кода.

3) Интерпретируем. Если связь очевидна — принимаем гипотезу. Если очевидно отсутствует — отвергаем. Если неочевидно — откладываем до CDA.

Для этого мы подготовили еще один датасет о службе доставки. В нём 500 заказов с признаками: регион, тип доставки, время, расстояние, рейтинг, количество товаров и та самая задержка, которая всех бесит.  Читаем датасет:

```python
df = pd.read_csv('data/df_logistics_2.csv')
df
```

Данные чистые, пропусков и выбросов нет. Мы сосредоточимся на связях.

### Структурные гипотезы

Структурные гипотезы — это те, которые можно проверить логикой и здравым смыслом без глубокой статистики. Они про то, как устроен мир: «Если А, то скорее всего Б».

#### Гипотеза №1: Чем больше расстояние, тем дольше время доставки

Звучит банально. Но давайте проверим.

```python
sns.scatterplot(data=df, x='расстояние_км', y='время_доставки_ч')
plt.show()
```

Что мы видим?

Точки образуют явное облако, вытянутое слева направо и снизу вверх. Чем больше расстояние, тем выше время. Да, есть разброс: одна и та же дистанция может быть преодолена за разное время (спасибо пробкам, погоде, человеческому фактору). Но тренд очевиден.

**Вердикт.** Гипотеза принята. Расстояние — драйвер времени доставки. Мы это и так знали, но приятно, что график не обманул.

#### Гипотеза №2: Экспресс-доставка быстрее стандартной

Тоже структурная логика: экспресс — значит быстрее.

```python
sns.boxplot(data=df, x='тип_доставки', y='время_доставки_ч')
plt.show()
```

Коробки говорят без слов. Медиана Экспресса — около 3 часов.Медиана Стандарта — все 8–9 часов. Бесконтактная — между ними, ближе к стандарту. Разброс экспресса меньше, чем у стандарта. Это логично: быстрые доставки — это сжатые сроки, мало времени на отклонения.  

**Вердикт.** Гипотеза Принята. Экспресс быстрее. Структурная гипотеза подтверждена. Бесконтактная — тоже быстрее стандарта, но не так сильно. Видимо, процедура подтверждения занимает время.

#### Гипотеза №3: Заказы с подарком получают более высокий рейтинг

Гипотеза из области человеческой психологии. Подарок вызывает радость → радость переносится на оценку сервиса.

```python
sns.boxplot(data=df, x='подарок', y='рейтинг_клиента')
plt.show()
```

Даже беглый взгляд говорит о многом. Без подарка — средний рейтинг около 3.3. С подарком — около 3.8. Разброс внутри групп перекрывается, но центры распределений явно смещены. 

**Вердикт.** Подарок работает. Люди любят приятные сюрпризы. Гипотеза структурная — мир устроен именно так. Но гипотеза требует статистического подтверждения на уровне CDA.

#### Другие структурные гипотезы

Мы рассмотрели три структурные гипотезы. Но это не значит, что их всего три. В нашем датасете, как и в любом реальном, можно сформулировать ещё десятки: «Чем больше товаров в заказе, тем выше вероятность подарка», «В центральном регионе чаще заказывают экспресс-доставку», «Заказы с бесконтактной доставкой имеют большую сумму». Некоторые из них окажутся тривиальными, некоторые — ложными, а некоторые — неожиданно ценными. Главное — не бояться их формулировать и проверять визуально. Одна строка кода — и вы либо получаете подтверждение интуиции, либо повод задуматься. Структурные гипотезы — это поле для творчества. Мы остановились на трёх, но вы вольны продолжить.

### Гипотезы о распределениях

Теперь мы смотрим не на связи «причина — следствие», а на поведение отдельных признаков. Нас интересует форма распределения: сколько у него «горбов», симметрично ли оно, есть ли скошенность.

Эти гипотезы нельзя проверить логикой. Только взглядом на гистограмму или kde-плотность. И, как мы увидим, взгляд может обмануть.

#### Гипотеза №4: Время доставки имеет бимодальное распределение

Логика: доставки делятся на «быстрые» (экспресс) и «медленные» (стандарт). Если смешать два распределения с разными центрами, мы должны увидеть два пика.

```python
sns.histplot(data=df, x='время_доставки_ч', bins=30, kde=True)
plt.show()
```

Видим, что предположение о бимодальности с треском провалилось.

**Вердикт.** Гипотеза отвергается. Ярко выраженная мода приходится на самые быстрые доставки.

#### Гипотеза №5: Рейтинг клиентов скошен вправо (есть правая асимметрия)

Гипотеза из области психологии: люди склонны ставить высокие оценки, а низкие — редко и только за серьёзные косяки.

```python
sns.histplot(data=df, x='рейтинг_клиента', bins=20, kde=True)
plt.show()
```

Смотрим. Большая масса оценок — от 2.5 до 4.5. Левый хвост (1.0 – 2.5) — длинный и тощий. Медиана явно смещена вправо.

**Вердикт.** Гипотеза принята. Рейтинг скошен влево. Большинство клиентов довольно, меньшинство — нет. Это согласуется с житейским опытом.

#### Гипотеза №6: Сумма заказа распределена равномерно

Вот здесь — ловушка. Многие думают: «суммы заказов бывают любые — от 200 до 5000 — и все одинаково вероятны». Ну… давайте проверим.

```python
sns.histplot(data=df, x='сумма_заказа_руб', bins=20, kde=True)
plt.show()
```

Вроде бы, все хорошо: распределение — не равномерное, но близко к нему. Все столбцы примерно одинаковой высоты. Есть две моды. Гипотеза подтвердилась?  Но подождите. В реальной жизни суммы заказов не бывают равномерными. Обычно есть мода в районе среднего чека (но никак не две моды), а очень большие и очень маленькие заказы — редки. Равномерное (или бимодальное) распределение — это… подозрительно.

**Вердикт.** Гипотеза отвергнута. Она визуально подтверждается, но является ложной по своей сути — так не бывает в реальности. Мы не верим, что суммы заказов распределены равномерно. Просим CDA? Нет. Здесь достаточно здравого смысла: если график показывает равномерность, а мы знаем, что в жизни так не бывает — гипотеза отвергается на этапе интерпретации. И если бы это были настоящие данные, мы бы заподозрили подтасовку.

Опровержение — не статистическое, а содержательное: мы не поверили графику, потому что он противоречит здравому смыслу. Это важный урок: график — не есть истина в последней инстанции.

#### Другие гипотезы о распределениях

Точно так же — мы не исчерпали список признаков. Расстояние, время, рейтинг, сумма, количество товаров, регион… да хоть цена доставки. У каждого из них есть своё распределение, и про каждое можно сформулировать гипотезу (и не одну): «Расстояние доставки распределено логнормально», «Количество товаров — пуассоновское», «Цена доставки имеет тяжелый правый хвост». Мы проверили три — и уже получили полезный урок про равномерное распределение (которому не поверили). Остальные признаки ждут своего часа. И это нормально: в учебном параграфе мы показываем метод, а не исчерпывающий перебор. В реальном проекте вы пройдётесь по всем числовым признакам хотя бы бегло — и найдёте свои сюрпризы.

### Комбинаторные гипотезы

Если структурные гипотезы — это про «как устроен мир», а гипотезы о распределениях — про «как устроен один признак», то комбинаторные гипотезы — про взаимодействие. Это уже не прямолинейная логика. Это детективная работа.

#### Гипотеза №7: Влияние расстояния на время доставки различается по регионам

Звучит логично: в центре дороги хорошие, пробки есть, но расстояния маленькие. На востоке — расстояния огромные, и дороги могут быть хуже. Время на километр будет разным.

```python
sns.scatterplot(data=df, x='расстояние_км', y='время_доставки_ч', hue='регион', alpha=0.6)
plt.show()
```

Видна общая для всех региноов тенденция. Но есть ли различия по регионам? Точки накладываются друг на друга и скрывают положение вещей. Давайте попробуем локализоваться только на ддва региеона: `'Центр'` и `'Восток'`:

```python
mask = (df['регион'] == 'Центр') | (df['регион'] == 'Восток')
sns.scatterplot(data=df.loc[mask], x='расстояние_км', y='время_доставки_ч', hue='регион', alpha=0.6)
plt.show()
```

Вывод очевиден: регион модерирует связь расстояния и времени. В центре — одна логика, на востоке — другая, и то, что для центра является доставкой на дальние расстояния, на востоке окзывается ближе самой короткой дистанции.


**Вердикт.** Гипотеза принята. Связь расстояния и времени не универсальна: регион является модератором.

#### Гипотеза №8: Зависимость рейтинга от задержки различается в зависимости от подарка

Хитрая гипотеза. Да, задержка снижает рейтинг — мы это знаем. Но может ли подарок смягчить падение? Психологический эффект: «меня задержали, но мне подарили шариковую ручку — и я уже не так зол».

```python
sns.boxplot(data=df, x='задержка', y='рейтинг_клиента', hue='подарок')
plt.show()
```

Сравниваем четыре коробочки.

- Нет задержки, нет подарка — рейтинг около 3.5.

- Нет задержки, есть подарок — заметно выше, ~4.0. (Это мы уже знаем.)

- Есть задержка, нет подарка — рейтинг падает, медиана около 2.7.

- Есть задержка, есть подарок — рейтин не падает так сильно! Медиана около 3.2. Подарок действительно «смягчает удар».

Между группами 3 и 4 есть перекрытие, но центры распределений различаются.

**Вердикт.** Гипотеза принята. Подарок при задержке действительно повышает рейтинг. Не до уровня «без задержки», но заметно выше, чем без подарка. Гипотеза о смягчающем эффекте подтверждена визуально.

#### Гипотеза №9 (коварная): Тип доставки и регион совместно влияют на задержку

Вот где начинается путаница. Мы уже знаем:

- экспресс-доставка быстрее (гипотеза №2),
- на востоке время доставки больше (гипотеза №7).

Вопрос: что первично? Может, на востоке просто реже заказывают экспресс, поэтому задержек больше? Или экспресс на востоке не спасает?

```python
sns.heatmap(pd.crosstab(df['регион'], df['тип_доставки'], normalize='index'), annot=True, cmap='YlOrRd')
plt.show()
```

Что мы видим?  


- Центр: 58% экспресс, 33% стандарт, 9% бесконтактная.

- Восток: 77% стандарт, 7% экспресс, 16% бесконтактная.

- Север и Юг — примерно поровну между стандартом и экспрессом.

Уже интересно. Восток — «царство стандарта». Теперь смотрим задержки по комбинациям:

```python
sns.heatmap(pd.crosstab([df['регион'], df['тип_доставки']], df['задержка'], normalize='index'), annot=True, cmap='RdYlGn_r')
plt.show()
```

Результат:

- Экспресс в центре: задержек не бывает.

- Стандарт на востоке: задержка в 61% случаев — катастрофа.

- Экспресс на востоке: задержка 50%. Тоже много, но меньше, чем 63%.

Вывод: Да, восток склонен к задержкам, но если на востоке выбрать экспресс, риск падает на 13%.

**Вердикт.** Гипотеза принята. Тип доставки и регион совместно влияют на задержку.

#### Другие комбинаторные гипотезы

Мы рассмотрели три комбинаторные гипотезы — про влияние региона, смягчение подарком и совместное действие типа доставки и региона. Но на этом возможности пересечения признаков не исчерпаны. Можно было бы проверить, как меняется связь суммы заказа и рейтинга в зависимости от наличия задержки, или как расстояние и количество товаров взаимодействуют в формировании времени доставки. Можно было бы построить трёхмерный scatter с цветом по региону и размером точек по подарку — чтобы увидеть кластеры, которые не видны в двумерных срезах. Или разбить данные на фасеты по типу доставки и посмотреть, как в каждом фасете работает связь расстояния и времени. Комбинаторные гипотезы — это пространство для исследования взаимодействий и скрытых паттернов. Мы остановились на трёх, но их могло быть двадцать — ограничение лишь в здравом смысле и времени аналитика. 

Главный урок: если вы видите связь между двумя переменными — всегда спрашивайте: «А не меняется ли она в зависимости от третьей?». Это и есть комбинаторное мышление.

Разведочный анализ — это технология, а не магия. У него есть шаги, их можно запомнить и выполнять по списку. Ты начинаешь с паники в глазах — «что с этими данными делать?».
